```python
import os
import zipfile
import random
import shutil
```

```python
BASE_PATH = "YOUR_PATH"

ZIP_PATH = os.path.join(BASE_PATH, "Legal_Docs.zip")
EXTRACT_PATH = os.path.join(BASE_PATH, "All_docs")
DOCS_PATH = os.path.join(EXTRACT_PATH, "Legal_Docs")

TRAIN_PATH = os.path.join(BASE_PATH, "Train")
TEST_PATH = os.path.join(BASE_PATH, "Test")
```

```python
if not os.path.exists(DOCS_PATH):
    with zipfile.ZipFile(ZIP_PATH, 'r') as zip_ref:
        zip_ref.extractall(EXTRACT_PATH)

print("ZIP extracted.")
```

```python
os.makedirs(TRAIN_PATH, exist_ok=True)
os.makedirs(TEST_PATH, exist_ok=True)

print("Train and Test folders ready.")
```

```python
all_files = []

for root, dirs, files in os.walk(DOCS_PATH):
    for f in files:
        if f.endswith(".txt"):
            all_files.append(os.path.join(root, f))

print("Total documents found:", len(all_files))
```

```python
random.seed(42)
random.shuffle(all_files)

split_index = int(0.8 * len(all_files))

train_files = all_files[:split_index]
test_files = all_files[split_index:]

print("Train:", len(train_files))
print("Test:", len(test_files))
```

```python
for src in train_files:
    shutil.copy(src, os.path.join(TRAIN_PATH, os.path.basename(src)))

for src in test_files:
    shutil.copy(src, os.path.join(TEST_PATH, os.path.basename(src)))

print("Train/Test split completed.")
```
