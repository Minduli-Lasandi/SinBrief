```python
# Paths
source_sample_dir = '/content/drive/MyDrive/SinBrief_Implementation/Dataset/Test/'
output_dir = '/content/drive/MyDrive/SinBrief_Implementation/Outputs/Falcon_Summaries/'
```

```python
def get_fragments(source, summary):
    source_tokens = source.split()
    summary_tokens = summary.split()
    fragments = []
    i = 0
    while i < len(summary_tokens):
        j = 0
        while i + j < len(summary_tokens):
            if " ".join(summary_tokens[i:i+j+1]) in source:
                j += 1
            else:
                break
        if j > 0:
            fragments.append(j)
            i += j
        else:
            i += 1
    return fragments

results = []
file_list = [f for f in os.listdir(output_dir) if f.endswith('.txt')]
print(f"Found {len(file_list)} summaries. Starting analysis...")

for file_name in file_list:
    # Remove "_summary"
    src_file = file_name.replace("_summary", "")

    src_path = os.path.join(source_sample_dir, src_file)
    sum_path = os.path.join(output_dir, file_name)

    if os.path.exists(src_path) and os.path.exists(sum_path):
        with open(src_path, 'r', encoding='utf-8') as f:
            src_text = f.read()
        with open(sum_path, 'r', encoding='utf-8') as f:
            sum_text = f.read()

        src_len = len(src_text.split())
        sum_len = len(sum_text.split())

        compression = sum_len / src_len if src_len > 0 else 0

        results.append({
            "File": file_name,
            "Src_Words": src_len,
            "Sum_Words": sum_len,
            "Compression_Ratio": round(compression, 4)
        })

df_final = pd.DataFrame(results)

output_csv = '/content/drive/MyDrive/SinBrief_Implementation/Evaluations/Falcon/Result_Sheets/compression_ratio.csv'
df_final.to_csv(output_csv, index=False)

print("Compression Ratio analysis complete.")
print("Compression Ratio - ",df_final["Compression_Ratio"].mean())
```
